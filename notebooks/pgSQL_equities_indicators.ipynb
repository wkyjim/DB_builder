```python
# ============================================================
# CONFIG
# ============================================================

password = "<LOCAL_DB_PASSWORD>"

DATABASE_URL = (
    f"postgresql://postgres:{password}"
    "@localhost:5432/us_equities_historical"
)

engine = create_engine(
    DATABASE_URL,

    # prevent stale connections
    pool_pre_ping=True,

    # recycle long-lived connections
    pool_recycle=3600,

    use_insertmanyvalues=True
)

RAW_TABLE = "public.us_equities"
INDICATOR_TABLE = "public.us_equities_indicators"

CHUNK_SIZE = 10000
```

```python
INDICATOR_COLUMNS = [
    "ma_5", "ma_20", "ma_50", "ma_100", "ma_200",
    "ema_12", "ema_26",
    "rsi_14",
    "macd", "macd_signal", "macd_hist",
    "atr_14",
    "volume_ma_20", "volume_ratio_20",
    "high_52w", "low_52w",
    "return_5d", "return_20d", "return_60d",
    "volatility_20d"
]
```

```python
# ============================================================
# INDICATOR CALCULATION
# ============================================================

def calculate_group(g: pd.DataFrame) -> pd.DataFrame:
    ticker = g["ticker"].iloc[0]

    g = g.sort_values("date").copy()
    g["ticker"] = ticker

    close = g["close"].astype(float)
    high = g["high"].astype(float)
    low = g["low"].astype(float)
    volume = g["volume"].astype(float)

    g["ma_5"] = close.rolling(5).mean()
    g["ma_20"] = close.rolling(20).mean()
    g["ma_50"] = close.rolling(50).mean()
    g["ma_100"] = close.rolling(100).mean()
    g["ma_200"] = close.rolling(200).mean()

    g["ema_12"] = close.ewm(span=12, adjust=False).mean()
    g["ema_26"] = close.ewm(span=26, adjust=False).mean()

    g["macd"] = g["ema_12"] - g["ema_26"]
    g["macd_signal"] = g["macd"].ewm(span=9, adjust=False).mean()
    g["macd_hist"] = g["macd"] - g["macd_signal"]

    delta = close.diff()
    gain = delta.clip(lower=0)
    loss = -delta.clip(upper=0)

    avg_gain = gain.rolling(14).mean()
    avg_loss = loss.rolling(14).mean()

    rs = avg_gain / avg_loss
    g["rsi_14"] = 100 - (100 / (1 + rs))

    true_range = pd.concat(
        [
            high - low,
            (high - close.shift()).abs(),
            (low - close.shift()).abs()
        ],
        axis=1
    ).max(axis=1)

    g["atr_14"] = true_range.rolling(14).mean()

    g["volume_ma_20"] = volume.rolling(20).mean()
    g["volume_ratio_20"] = volume / g["volume_ma_20"]

    g["high_52w"] = high.rolling(252).max()
    g["low_52w"] = low.rolling(252).min()

    g["return_5d"] = close.pct_change(5) * 100
    g["return_20d"] = close.pct_change(20) * 100
    g["return_60d"] = close.pct_change(60) * 100

    g["volatility_20d"] = (
        close.pct_change()
        .rolling(20)
        .std()
        * np.sqrt(252)
        * 100
    )

    return g[["date", "ticker"] + INDICATOR_COLUMNS]


# ============================================================
# BULK UPSERT WITH PROGRESS
# ============================================================

def bulk_upsert_with_progress(df: pd.DataFrame, chunk_size: int = CHUNK_SIZE):
    start_time = time.time()

    df = df.reset_index(drop=True)

    required_cols = ["date", "ticker"] + INDICATOR_COLUMNS
    missing_cols = [c for c in required_cols if c not in df.columns]

    if missing_cols:
        raise ValueError(f"Missing columns: {missing_cols}")

    df = df[required_cols]
    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.where(pd.notnull(df), None)

    sql = f"""
    INSERT INTO {INDICATOR_TABLE} (
        {", ".join(required_cols)}
    )
    VALUES (
        {", ".join([":" + c for c in required_cols])}
    )
    ON CONFLICT (ticker, date)
    DO UPDATE SET
        {", ".join([f"{c} = EXCLUDED.{c}" for c in INDICATOR_COLUMNS])};
    """

    records = df.to_dict(orient="records")
    total_records = len(records)

    if total_records == 0:
        print("No rows to upsert.")
        return

    print(f"Rows to upsert: {total_records:,}")
    print(f"Chunk size: {chunk_size:,}")
    print("-" * 70)

    with engine.begin() as conn:
        for i in range(0, total_records, chunk_size):
            chunk = records[i:i + chunk_size]
            conn.execute(text(sql), chunk)

            done = min(i + chunk_size, total_records)
            pct = done / total_records * 100
            elapsed = time.time() - start_time

            print(
                f"Upserted {done:,}/{total_records:,} rows "
                f"({pct:.1f}%) | "
                f"Elapsed: {elapsed / 60:.1f} min"
            )

    print("-" * 70)
    print("Bulk upsert finished.")


# ============================================================
# FULL BACKFILL REMAINING TICKERS ONLY
# ============================================================

def full_backfill_remaining_tickers():
    start_time = time.time()

    print("=" * 70)
    print("Starting full backfill for remaining tickers")
    print("=" * 70)

    raw_ticker_count = pd.read_sql(
        f"""
        SELECT COUNT(DISTINCT ticker) AS cnt
        FROM {RAW_TABLE}
        """,
        engine
    )["cnt"].iloc[0]

    processed_ticker_count = pd.read_sql(
        f"""
        SELECT COUNT(DISTINCT ticker) AS cnt
        FROM {INDICATOR_TABLE}
        """,
        engine
    )["cnt"].iloc[0]

    print(f"Raw tickers: {raw_ticker_count:,}")
    print(f"Already processed tickers: {processed_ticker_count:,}")
    print(f"Remaining tickers: {raw_ticker_count - processed_ticker_count:,}")
    print("-" * 70)

    df = pd.read_sql(
        f"""
        SELECT
            date,
            ticker,
            open,
            high,
            low,
            close,
            volume
        FROM {RAW_TABLE}
        WHERE ticker NOT IN (
            SELECT DISTINCT ticker
            FROM {INDICATOR_TABLE}
        )
        ORDER BY ticker, date
        """,
        engine
    )

    if df.empty:
        print("No remaining tickers to backfill.")
        return

    print(f"Loaded rows: {len(df):,}")
    print(f"Loaded remaining tickers: {df['ticker'].nunique():,}")
    print("-" * 70)

    results = []
    grouped = list(df.groupby("ticker", group_keys=False))
    total_tickers = len(grouped)

    print("Calculating indicators...")
    print("-" * 70)

    for idx, (ticker, group) in enumerate(grouped, start=1):
        try:
            result = calculate_group(group).reset_index(drop=True)
            result["ticker"] = ticker
            results.append(result)

            if idx % 25 == 0 or idx == total_tickers:
                elapsed = time.time() - start_time
                print(
                    f"[{idx:,}/{total_tickers:,}] "
                    f"Calculated {ticker} | "
                    f"Elapsed: {elapsed / 60:.1f} min"
                )

        except Exception as e:
            print(f"ERROR calculating {ticker}: {e}")

    if not results:
        print("No indicator results were calculated.")
        return

    print("-" * 70)
    print("Combining calculated results...")

    result_df = pd.concat(results, ignore_index=True)

    print(f"Calculated indicator rows: {len(result_df):,}")
    print(f"Calculated tickers: {result_df['ticker'].nunique():,}")
    print("-" * 70)

    print("Starting bulk upsert...")

    bulk_upsert_with_progress(result_df, chunk_size=CHUNK_SIZE)

    elapsed = time.time() - start_time

    print("=" * 70)
    print("Remaining ticker full backfill complete.")
    print(f"Total elapsed: {elapsed / 60:.1f} minutes")
    print("=" * 70)
```

```python
# ============================================================
# RUN
# ============================================================

if __name__ == "__main__":
    full_backfill_remaining_tickers()
```

```python
# ============================================================
# INDICATOR CALCULATION
# ============================================================

def calculate_group(g: pd.DataFrame) -> pd.DataFrame:
    ticker = g["ticker"].iloc[0]

    g = g.sort_values("date").copy()
    g["ticker"] = ticker

    close = g["close"].astype(float)
    high = g["high"].astype(float)
    low = g["low"].astype(float)
    volume = g["volume"].astype(float)

    g["ma_5"] = close.rolling(5).mean()
    g["ma_20"] = close.rolling(20).mean()
    g["ma_50"] = close.rolling(50).mean()
    g["ma_100"] = close.rolling(100).mean()
    g["ma_200"] = close.rolling(200).mean()

    g["ema_12"] = close.ewm(span=12, adjust=False).mean()
    g["ema_26"] = close.ewm(span=26, adjust=False).mean()

    g["macd"] = g["ema_12"] - g["ema_26"]
    g["macd_signal"] = g["macd"].ewm(span=9, adjust=False).mean()
    g["macd_hist"] = g["macd"] - g["macd_signal"]

    delta = close.diff()
    gain = delta.clip(lower=0)
    loss = -delta.clip(upper=0)

    avg_gain = gain.rolling(14).mean()
    avg_loss = loss.rolling(14).mean()

    rs = avg_gain / avg_loss
    g["rsi_14"] = 100 - (100 / (1 + rs))

    true_range = pd.concat(
        [
            high - low,
            (high - close.shift()).abs(),
            (low - close.shift()).abs()
        ],
        axis=1
    ).max(axis=1)

    g["atr_14"] = true_range.rolling(14).mean()

    g["volume_ma_20"] = volume.rolling(20).mean()
    g["volume_ratio_20"] = volume / g["volume_ma_20"]

    g["high_52w"] = high.rolling(252).max()
    g["low_52w"] = low.rolling(252).min()

    g["return_5d"] = close.pct_change(5) * 100
    g["return_20d"] = close.pct_change(20) * 100
    g["return_60d"] = close.pct_change(60) * 100

    g["volatility_20d"] = (
        close.pct_change()
        .rolling(20)
        .std()
        * np.sqrt(252)
        * 100
    )

    return g[["date", "ticker"] + INDICATOR_COLUMNS]


# ============================================================
# BULK UPSERT
# ============================================================

def bulk_upsert_with_progress(df: pd.DataFrame, chunk_size: int = CHUNK_SIZE):
    start_time = time.time()

    df = df.reset_index(drop=True)

    required_cols = ["date", "ticker"] + INDICATOR_COLUMNS
    missing_cols = [c for c in required_cols if c not in df.columns]

    if missing_cols:
        raise ValueError(f"Missing columns: {missing_cols}")

    df = df[required_cols]
    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.where(pd.notnull(df), None)

    records = df.to_dict(orient="records")
    total_records = len(records)

    if total_records == 0:
        print("No rows to upsert.")
        return

    sql = f"""
    INSERT INTO {INDICATOR_TABLE} (
        {", ".join(required_cols)}
    )
    VALUES (
        {", ".join([":" + c for c in required_cols])}
    )
    ON CONFLICT (ticker, date)
    DO UPDATE SET
        {", ".join([f"{c} = EXCLUDED.{c}" for c in INDICATOR_COLUMNS])};
    """

    print(f"Rows to upsert: {total_records:,}")
    print(f"Chunk size: {chunk_size:,}")
    print("-" * 70)

    with engine.begin() as conn:
        for i in range(0, total_records, chunk_size):
            chunk = records[i:i + chunk_size]
            conn.execute(text(sql), chunk)

            done = min(i + chunk_size, total_records)
            pct = done / total_records * 100
            elapsed = time.time() - start_time

            print(
                f"Upserted {done:,}/{total_records:,} rows "
                f"({pct:.1f}%) | "
                f"Elapsed: {elapsed / 60:.1f} min"
            )

    print("-" * 70)
    print("Bulk upsert finished.")


# ============================================================
# DAILY UPDATE MISSING/LATEST INDICATOR DATES
# ============================================================

def daily_update_missing_indicators(lookback_rows: int = LOOKBACK_ROWS):
    start_time = time.time()

    print("=" * 70)
    print("Starting daily update for missing/latest indicator dates")
    print("=" * 70)

    latest_raw = pd.read_sql(
        f"""
        SELECT
            ticker,
            MAX(date) AS latest_raw_date
        FROM {RAW_TABLE}
        GROUP BY ticker
        """,
        engine
    )

    latest_ind = pd.read_sql(
        f"""
        SELECT
            ticker,
            MAX(date) AS latest_indicator_date
        FROM {INDICATOR_TABLE}
        GROUP BY ticker
        """,
        engine
    )

    status = latest_raw.merge(
        latest_ind,
        on="ticker",
        how="left"
    )

    status_to_update = status[
        status["latest_indicator_date"].isna()
        | (status["latest_raw_date"] > status["latest_indicator_date"])
    ].copy()

    if status_to_update.empty:
        print("No tickers need indicator updates.")
        return

    tickers_to_update = status_to_update["ticker"].tolist()

    print(f"Raw tickers: {len(latest_raw):,}")
    print(f"Tickers needing update: {len(tickers_to_update):,}")
    print("-" * 70)

    df = pd.read_sql(
        f"""
        WITH ranked AS (
            SELECT
                date,
                ticker,
                open,
                high,
                low,
                close,
                volume,
                ROW_NUMBER() OVER (
                    PARTITION BY ticker
                    ORDER BY date DESC
                ) AS rn
            FROM {RAW_TABLE}
            WHERE ticker = ANY(%(tickers)s)
        )
        SELECT
            date,
            ticker,
            open,
            high,
            low,
            close,
            volume
        FROM ranked
        WHERE rn <= %(lookback_rows)s
        ORDER BY ticker, date
        """,
        engine,
        params={
            "tickers": tickers_to_update,
            "lookback_rows": lookback_rows
        }
    )

    if df.empty:
        print("No rows loaded for update.")
        return

    print(f"Loaded rows: {len(df):,}")
    print(f"Loaded tickers: {df['ticker'].nunique():,}")
    print("-" * 70)

    latest_indicator_map = dict(
        zip(
            status_to_update["ticker"],
            status_to_update["latest_indicator_date"]
        )
    )

    results = []
    grouped = list(df.groupby("ticker", group_keys=False))
    total_tickers = len(grouped)

    print("Calculating indicators...")
    print("-" * 70)

    for idx, (ticker, group) in enumerate(grouped, start=1):
        try:
            result = calculate_group(group).reset_index(drop=True)
            result["ticker"] = ticker

            latest_indicator_date = latest_indicator_map.get(ticker)

            if pd.notnull(latest_indicator_date):
                result = result[result["date"] > latest_indicator_date]

            if not result.empty:
                results.append(result)

            if idx % 25 == 0 or idx == total_tickers:
                queued_rows = sum(len(r) for r in results)
                elapsed = time.time() - start_time

                print(
                    f"[{idx:,}/{total_tickers:,}] "
                    f"Processed {ticker} | "
                    f"Rows queued: {queued_rows:,} | "
                    f"Elapsed: {elapsed / 60:.1f} min"
                )

        except Exception as e:
            print(f"ERROR calculating {ticker}: {e}")

    if not results:
        print("No new indicator rows to upsert.")
        return

    result_df = pd.concat(results, ignore_index=True)

    print("-" * 70)
    print(f"New indicator rows to upsert: {len(result_df):,}")
    print(f"Tickers updated: {result_df['ticker'].nunique():,}")
    print("-" * 70)

    bulk_upsert_with_progress(result_df, chunk_size=CHUNK_SIZE)

    elapsed = time.time() - start_time

    print("=" * 70)
    print("Daily indicator update complete.")
    print(f"Total elapsed: {elapsed / 60:.1f} minutes")
    print("=" * 70)
```

```python
import time
import numpy as np
import pandas as pd
from sqlalchemy import create_engine, text
```

```python
# ============================================================
# CONFIG
# ============================================================

local_password = "<LOCAL_DB_PASSWORD>"
neon_password = "<NEON_DB_PASSWORD>"

LOCAL_DATABASE_URL = (
    f"postgresql://postgres:{local_password}"
    "@localhost:5432/us_equities_historical"
)

NEON_DATABASE_URL = (
    f"postgresql://neondb_owner:{neon_password}"
    "@ep-aged-moon-ao3o4z0j-pooler.c-2.ap-southeast-1.aws.neon.tech/"
    "neondb?sslmode=require&channel_binding=require"
)

local_engine = create_engine(
    LOCAL_DATABASE_URL,
    pool_pre_ping=True,
    pool_recycle=3600
)

neon_engine = create_engine(
    NEON_DATABASE_URL,
    pool_pre_ping=True,
    pool_recycle=3600
)

RAW_TABLE = "public.us_equities"
INDICATOR_TABLE = "public.us_equities_indicators"

CHUNK_SIZE = 5000

# Re-upload a small recent window to avoid missing late/changed rows.
# Because we use UPSERT, duplicates are safe.
OVERLAP_DAYS = 5
```

```python
# ============================================================
# SETUP NEON SYNC STATE
# ============================================================

def setup_sync_state():
    sql = """
    CREATE TABLE IF NOT EXISTS public.sync_state (
        table_name text PRIMARY KEY,
        latest_date date,
        updated_at timestamp DEFAULT now()
    );
    """

    with neon_engine.begin() as conn:
        conn.execute(text(sql))

    print("sync_state table ready.")


def get_sync_latest_date(table_name: str):
    sql = """
    SELECT latest_date
    FROM public.sync_state
    WHERE table_name = :table_name
    """

    df = pd.read_sql(
        text(sql),
        neon_engine,
        params={"table_name": table_name}
    )

    if df.empty or pd.isna(df["latest_date"].iloc[0]):
        return None

    return df["latest_date"].iloc[0]


def update_sync_latest_date(table_name: str, latest_date):
    sql = """
    INSERT INTO public.sync_state (
        table_name,
        latest_date,
        updated_at
    )
    VALUES (
        :table_name,
        :latest_date,
        now()
    )
    ON CONFLICT (table_name)
    DO UPDATE SET
        latest_date = EXCLUDED.latest_date,
        updated_at = now();
    """

    with neon_engine.begin() as conn:
        conn.execute(
            text(sql),
            {
                "table_name": table_name,
                "latest_date": latest_date
            }
        )


# ============================================================
# LOCAL DATA FETCH
# ============================================================

def fetch_local_rows_after_date(local_table: str, latest_date):
    if latest_date is None:
        print(f"No sync date found for {local_table}. Fetching all rows.")

        sql = f"""
        SELECT *
        FROM {local_table}
        ORDER BY date, ticker
        """

        return pd.read_sql(sql, local_engine)

    print(
        f"Fetching rows from {local_table} "
        f"where date >= {latest_date} - {OVERLAP_DAYS} days"
    )

    sql = f"""
    SELECT *
    FROM {local_table}
    WHERE date >= (%(latest_date)s::date - INTERVAL '{OVERLAP_DAYS} days')
    ORDER BY date, ticker
    """

    return pd.read_sql(
        sql,
        local_engine,
        params={"latest_date": latest_date}
    )


# ============================================================
# BULK UPSERT TO NEON
# ============================================================

def bulk_upsert_to_neon(
    df: pd.DataFrame,
    neon_table: str,
    conflict_cols: list[str],
    chunk_size: int = CHUNK_SIZE
):
    if df.empty:
        print(f"No rows to upload for {neon_table}.")
        return None

    df = df.reset_index(drop=True)
    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.where(pd.notnull(df), None)

    columns = df.columns.tolist()

    update_cols = [
        c for c in columns
        if c not in conflict_cols
    ]

    sql = f"""
    INSERT INTO {neon_table} (
        {", ".join(columns)}
    )
    VALUES (
        {", ".join([":" + c for c in columns])}
    )
    ON CONFLICT ({", ".join(conflict_cols)})
    DO UPDATE SET
        {", ".join([f"{c} = EXCLUDED.{c}" for c in update_cols])};
    """

    records = df.to_dict(orient="records")
    total = len(records)

    print("-" * 70)
    print(f"Uploading {total:,} rows to {neon_table}")
    print(f"Chunk size: {chunk_size:,}")
    print("-" * 70)

    start_time = time.time()

    with neon_engine.begin() as conn:
        for i in range(0, total, chunk_size):
            chunk = records[i:i + chunk_size]
            conn.execute(text(sql), chunk)

            done = min(i + chunk_size, total)
            pct = done / total * 100
            elapsed = time.time() - start_time

            print(
                f"{neon_table}: {done:,}/{total:,} rows "
                f"({pct:.1f}%) | elapsed {elapsed / 60:.1f} min"
            )

    latest_uploaded_date = df["date"].max()

    print(f"Finished uploading {neon_table}")
    print(f"Latest uploaded date: {latest_uploaded_date}")

    return latest_uploaded_date


# ============================================================
# DAILY SYNC FOR ONE TABLE
# ============================================================

def sync_one_table(table_name: str):
    print("=" * 70)
    print(f"Syncing {table_name}")
    print("=" * 70)

    latest_date = get_sync_latest_date(table_name)

    print(f"Neon sync_state latest date for {table_name}: {latest_date}")

    df = fetch_local_rows_after_date(
        local_table=table_name,
        latest_date=latest_date
    )

    print(f"Rows fetched from local {table_name}: {len(df):,}")

    if df.empty:
        print(f"No new rows for {table_name}.")
        return

    latest_uploaded_date = bulk_upsert_to_neon(
        df=df,
        neon_table=table_name,
        conflict_cols=["ticker", "date"],
        chunk_size=CHUNK_SIZE
    )

    if latest_uploaded_date is not None:
        update_sync_latest_date(
            table_name=table_name,
            latest_date=latest_uploaded_date
        )

        print(
            f"Updated sync_state for {table_name} "
            f"to {latest_uploaded_date}"
        )


# ============================================================
# DAILY SYNC MAIN
# ============================================================

def daily_bulk_sync_to_neon():
    start_time = time.time()

    print("=" * 70)
    print("Starting daily bulk sync to Neon")
    print("=" * 70)

    setup_sync_state()

    sync_one_table(RAW_TABLE)
    sync_one_table(INDICATOR_TABLE)

    elapsed = time.time() - start_time

    print("=" * 70)
    print("Daily bulk sync to Neon complete.")
    print(f"Total elapsed: {elapsed / 60:.1f} minutes")
    print("=" * 70)
```

```python
# ============================================================
# RUN
# ============================================================

if __name__ == "__main__":
    daily_bulk_sync_to_neon()
```

```python
import time
import numpy as np
import pandas as pd
from sqlalchemy import create_engine, text


# ============================================================
# CONFIG
# ============================================================

local_password = "<LOCAL_DB_PASSWORD>"
neon_password = "<NEON_DB_PASSWORD>"

CUSTOM_DATE = "2026-05-08"

LOCAL_DATABASE_URL = (
    f"postgresql://postgres:{local_password}"
    "@localhost:5432/us_equities_historical"
)

NEON_DATABASE_URL = (
    f"postgresql://neondb_owner:{neon_password}"
    "@ep-aged-moon-ao3o4z0j-pooler.c-2.ap-southeast-1.aws.neon.tech/"
    "neondb?sslmode=require&channel_binding=require"
)

local_engine = create_engine(
    LOCAL_DATABASE_URL,
    pool_pre_ping=True,
    pool_recycle=3600
)

neon_engine = create_engine(
    NEON_DATABASE_URL,
    pool_pre_ping=True,
    pool_recycle=3600
)

RAW_TABLE = "public.us_equities"
INDICATOR_TABLE = "public.us_equities_indicators"

CHUNK_SIZE = 5000

RAW_NUMERIC_COLUMNS = [
    "open", "high", "low", "close", "change", "pct_chg", "prev_close",
    "turnover", "volume", "mkt_cap", "ytd_pct_chg", "pe_ttm",
    "amplitude", "turnover_rate"
]

INDICATOR_NUMERIC_COLUMNS = [
    "ma_5", "ma_20", "ma_50", "ma_100", "ma_200",
    "ema_12", "ema_26", "rsi_14",
    "macd", "macd_signal", "macd_hist",
    "atr_14",
    "volume_ma_20", "volume_ratio_20",
    "high_52w", "low_52w",
    "return_5d", "return_20d", "return_60d",
    "volatility_20d"
]


# ============================================================
# SETUP NEON
# ============================================================

def setup_neon_required_objects():
    sql = """
    CREATE TABLE IF NOT EXISTS public.sync_state (
        table_name text PRIMARY KEY,
        latest_date date,
        updated_at timestamp DEFAULT now()
    );

    CREATE UNIQUE INDEX IF NOT EXISTS idx_us_equities_ticker_date
    ON public.us_equities (ticker, date);

    CREATE UNIQUE INDEX IF NOT EXISTS idx_us_equities_indicators_ticker_date
    ON public.us_equities_indicators (ticker, date);

    -- Archived notebook: staging cleanup removed; temp tables are used by scripts.
    -- Archived notebook: staging cleanup removed; temp tables are used by scripts.
    """

    with neon_engine.begin() as conn:
        conn.execute(text(sql))

    print("Neon sync_state, indexes, and old staging cleanup complete.")


def update_sync_state(table_name: str, latest_date: str):
    sql = """
    INSERT INTO public.sync_state (
        table_name,
        latest_date,
        updated_at
    )
    VALUES (
        :table_name,
        :latest_date,
        now()
    )
    ON CONFLICT (table_name)
    DO UPDATE SET
        latest_date = EXCLUDED.latest_date,
        updated_at = now();
    """

    with neon_engine.begin() as conn:
        conn.execute(
            text(sql),
            {
                "table_name": table_name,
                "latest_date": latest_date
            }
        )


# ============================================================
# LOCAL FETCH
# ============================================================

def fetch_local_rows_for_date(table_name: str, target_date: str) -> pd.DataFrame:
    sql = f"""
    SELECT *
    FROM {table_name}
    WHERE date = %(target_date)s
    ORDER BY ticker
    """

    df = pd.read_sql(
        sql,
        local_engine,
        params={"target_date": target_date}
    )

    print(f"{table_name}: fetched {len(df):,} rows for {target_date}")
    return df


# ============================================================
# DATA CLEANING
# ============================================================

def clean_dataframe_for_target(df: pd.DataFrame, table_name: str) -> pd.DataFrame:
    df = df.copy()

    if "date" in df.columns:
        df["date"] = pd.to_datetime(df["date"]).dt.date

    numeric_cols = (
        RAW_NUMERIC_COLUMNS
        if table_name == RAW_TABLE
        else INDICATOR_NUMERIC_COLUMNS
    )

    for col in numeric_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

    # Ensure ticker/name/market are strings when present
    for col in ["ticker", "name", "market"]:
        if col in df.columns:
            df[col] = df[col].astype("string")

    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.where(pd.notnull(df), None)

    return df


# ============================================================
# TEMP STAGING BULK UPSERT
# ============================================================

def bulk_temp_staging_upsert_to_neon(
    df: pd.DataFrame,
    target_table: str,
    conflict_cols: list[str],
    chunk_size: int = CHUNK_SIZE
):
    if df.empty:
        print(f"No rows to upload for {target_table}.")
        return False

    start_time = time.time()

    df = clean_dataframe_for_target(df, target_table)

    columns = df.columns.tolist()
    update_cols = [c for c in columns if c not in conflict_cols]

    temp_table = f"temp_{target_table.split('.')[-1]}"
    column_list = ", ".join(columns)

    merge_sql = f"""
    INSERT INTO {target_table} (
        {column_list}
    )
    SELECT
        {column_list}
    FROM {temp_table}
    ON CONFLICT ({", ".join(conflict_cols)})
    DO UPDATE SET
        {", ".join([f"{c} = EXCLUDED.{c}" for c in update_cols])};
    """

    print("-" * 70)
    print(f"Bulk uploading {len(df):,} rows to {target_table}")
    print("Using TEMP staging table with same schema as target.")
    print("-" * 70)

    with neon_engine.begin() as conn:
        conn.execute(text(f"""
            CREATE TEMP TABLE {temp_table}
            (LIKE {target_table} INCLUDING DEFAULTS)
            ON COMMIT DROP;
        """))

        df.to_sql(
            name=temp_table,
            con=conn,
            if_exists="append",
            index=False,
            chunksize=chunk_size,
            method="multi"
        )

        conn.execute(text(merge_sql))

    elapsed = time.time() - start_time

    print(f"Finished bulk temp-staging upsert to {target_table}")
    print(f"Elapsed: {elapsed / 60:.1f} min")

    return True


# ============================================================
# FIRST-TIME CUSTOM DATE UPLOAD
# ============================================================

def first_time_bulk_upload_custom_date(target_date: str):
    start_time = time.time()

    print("=" * 70)
    print(f"Starting first-time bulk upload for {target_date}")
    print("=" * 70)

    setup_neon_required_objects()

    for table_name in [RAW_TABLE, INDICATOR_TABLE]:
        print("=" * 70)
        print(f"Processing {table_name}")
        print("=" * 70)

        df = fetch_local_rows_for_date(table_name, target_date)

        uploaded = bulk_temp_staging_upsert_to_neon(
            df=df,
            target_table=table_name,
            conflict_cols=["ticker", "date"],
            chunk_size=CHUNK_SIZE
        )

        if uploaded:
            update_sync_state(
                table_name=table_name,
                latest_date=target_date
            )

            print(f"sync_state updated: {table_name} = {target_date}")

    elapsed = time.time() - start_time

    print("=" * 70)
    print("First-time custom-date bulk upload complete.")
    print(f"Total elapsed: {elapsed / 60:.1f} min")
    print("=" * 70)
```

```python
import time
import numpy as np
import pandas as pd
from sqlalchemy import create_engine, text


# ============================================================
# CONFIG
# ============================================================

local_password = "<LOCAL_DB_PASSWORD>"
neon_password = "<NEON_DB_PASSWORD>"

LOCAL_DATABASE_URL = (
    f"postgresql://postgres:{local_password}"
    "@localhost:5432/us_equities_historical"
)

NEON_DATABASE_URL = (
    f"postgresql://neondb_owner:{neon_password}"
    "@ep-aged-moon-ao3o4z0j-pooler.c-2.ap-southeast-1.aws.neon.tech/"
    "neondb?sslmode=require&channel_binding=require"
)

local_engine = create_engine(LOCAL_DATABASE_URL, pool_pre_ping=True, pool_recycle=3600)
neon_engine = create_engine(NEON_DATABASE_URL, pool_pre_ping=True, pool_recycle=3600)

RAW_TABLE = "public.us_equities"
INDICATOR_TABLE = "public.us_equities_indicators"

CHUNK_SIZE = 5000
OVERLAP_DAYS = 5

RAW_NUMERIC_COLUMNS = [
    "open", "high", "low", "close", "change", "pct_chg", "prev_close",
    "turnover", "volume", "mkt_cap", "ytd_pct_chg", "pe_ttm",
    "amplitude", "turnover_rate"
]

INDICATOR_NUMERIC_COLUMNS = [
    "ma_5", "ma_20", "ma_50", "ma_100", "ma_200",
    "ema_12", "ema_26", "rsi_14",
    "macd", "macd_signal", "macd_hist",
    "atr_14",
    "volume_ma_20", "volume_ratio_20",
    "high_52w", "low_52w",
    "return_5d", "return_20d", "return_60d",
    "volatility_20d"
]


# ============================================================
# NEON SETUP / SYNC STATE
# ============================================================

def setup_neon_required_objects():
    sql = """
    CREATE TABLE IF NOT EXISTS public.sync_state (
        table_name text PRIMARY KEY,
        latest_date date,
        updated_at timestamp DEFAULT now()
    );

    CREATE UNIQUE INDEX IF NOT EXISTS idx_us_equities_ticker_date
    ON public.us_equities (ticker, date);

    CREATE UNIQUE INDEX IF NOT EXISTS idx_us_equities_indicators_ticker_date
    ON public.us_equities_indicators (ticker, date);

    -- Archived notebook: staging cleanup removed; temp tables are used by scripts.
    -- Archived notebook: staging cleanup removed; temp tables are used by scripts.
    """

    with neon_engine.begin() as conn:
        conn.execute(text(sql))

    print("Neon required objects ready.")


def get_sync_latest_date(table_name: str):
    sql = """
    SELECT latest_date
    FROM public.sync_state
    WHERE table_name = :table_name
    """

    df = pd.read_sql(
        text(sql),
        neon_engine,
        params={"table_name": table_name}
    )

    if df.empty or pd.isna(df["latest_date"].iloc[0]):
        raise ValueError(
            f"No sync_state found for {table_name}. "
            "Run the first-time custom-date upload first."
        )

    return df["latest_date"].iloc[0]


def update_sync_state(table_name: str, latest_date):
    sql = """
    INSERT INTO public.sync_state (
        table_name,
        latest_date,
        updated_at
    )
    VALUES (
        :table_name,
        :latest_date,
        now()
    )
    ON CONFLICT (table_name)
    DO UPDATE SET
        latest_date = EXCLUDED.latest_date,
        updated_at = now();
    """

    with neon_engine.begin() as conn:
        conn.execute(
            text(sql),
            {
                "table_name": table_name,
                "latest_date": latest_date
            }
        )


# ============================================================
# LOCAL FETCH
# ============================================================

def fetch_local_rows_for_daily_sync(
    table_name: str,
    latest_date,
    overlap_days: int = OVERLAP_DAYS
) -> pd.DataFrame:
    sql = f"""
    SELECT *
    FROM {table_name}
    WHERE date >= (%(latest_date)s::date - INTERVAL '{overlap_days} days')
    ORDER BY date, ticker
    """

    df = pd.read_sql(
        sql,
        local_engine,
        params={"latest_date": latest_date}
    )

    print(
        f"{table_name}: fetched {len(df):,} rows "
        f"from local where date >= {latest_date} - {overlap_days} days"
    )

    return df


# ============================================================
# DATA CLEANING
# ============================================================

def clean_dataframe_for_target(df: pd.DataFrame, table_name: str) -> pd.DataFrame:
    df = df.copy()

    if "date" in df.columns:
        df["date"] = pd.to_datetime(df["date"]).dt.date

    numeric_cols = (
        RAW_NUMERIC_COLUMNS
        if table_name == RAW_TABLE
        else INDICATOR_NUMERIC_COLUMNS
    )

    for col in numeric_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

    for col in ["ticker", "name", "market"]:
        if col in df.columns:
            df[col] = df[col].astype("string")

    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.where(pd.notnull(df), None)

    return df


# ============================================================
# TEMP STAGING BULK UPSERT
# ============================================================

def bulk_temp_staging_upsert_to_neon(
    df: pd.DataFrame,
    target_table: str,
    conflict_cols: list[str],
    chunk_size: int = CHUNK_SIZE
):
    if df.empty:
        print(f"No rows to upload for {target_table}.")
        return None

    start_time = time.time()

    df = clean_dataframe_for_target(df, target_table)

    columns = df.columns.tolist()
    update_cols = [c for c in columns if c not in conflict_cols]

    temp_table = f"temp_{target_table.split('.')[-1]}"
    column_list = ", ".join(columns)

    merge_sql = f"""
    INSERT INTO {target_table} (
        {column_list}
    )
    SELECT
        {column_list}
    FROM {temp_table}
    ON CONFLICT ({", ".join(conflict_cols)})
    DO UPDATE SET
        {", ".join([f"{c} = EXCLUDED.{c}" for c in update_cols])};
    """

    print("-" * 70)
    print(f"Bulk uploading {len(df):,} rows to {target_table}")
    print("Using TEMP staging table with same schema as target.")
    print("-" * 70)

    with neon_engine.begin() as conn:
        conn.execute(text(f"""
            CREATE TEMP TABLE {temp_table}
            (LIKE {target_table} INCLUDING DEFAULTS)
            ON COMMIT DROP;
        """))

        df.to_sql(
            name=temp_table,
            con=conn,
            if_exists="append",
            index=False,
            chunksize=chunk_size,
            method="multi"
        )

        conn.execute(text(merge_sql))

    latest_uploaded_date = df["date"].max()
    elapsed = time.time() - start_time

    print(f"Finished bulk temp-staging upsert to {target_table}")
    print(f"Latest uploaded date: {latest_uploaded_date}")
    print(f"Elapsed: {elapsed / 60:.1f} min")

    return latest_uploaded_date


# ============================================================
# DAILY SYNC
# ============================================================

def daily_bulk_sync_to_neon():
    start_time = time.time()

    print("=" * 70)
    print("Starting daily bulk sync to Neon")
    print("=" * 70)

    setup_neon_required_objects()

    for table_name in [RAW_TABLE, INDICATOR_TABLE]:
        print("=" * 70)
        print(f"Processing {table_name}")
        print("=" * 70)

        latest_date = get_sync_latest_date(table_name)
        print(f"sync_state latest date for {table_name}: {latest_date}")

        df = fetch_local_rows_for_daily_sync(
            table_name=table_name,
            latest_date=latest_date,
            overlap_days=OVERLAP_DAYS
        )

        latest_uploaded_date = bulk_temp_staging_upsert_to_neon(
            df=df,
            target_table=table_name,
            conflict_cols=["ticker", "date"],
            chunk_size=CHUNK_SIZE
        )

        if latest_uploaded_date is not None:
            update_sync_state(table_name, latest_uploaded_date)
            print(f"sync_state updated: {table_name} = {latest_uploaded_date}")

    elapsed = time.time() - start_time

    print("=" * 70)
    print("Daily bulk sync to Neon complete.")
    print(f"Total elapsed: {elapsed / 60:.1f} min")
    print("=" * 70)
```

```python
#### ============================================================
# RUN
# ============================================================

if __name__ == "__main__":
    daily_bulk_sync_to_neon()
```
