```python
import requests
import pandas as pd
import time
import random
from datetime import datetime, timedelta
from sqlalchemy import create_engine, MetaData, Table
from sqlalchemy.dialects.postgresql import insert as pg_insert
```

```python
# --- 配置 ---
password = "<LOCAL_DB_PASSWORD>"
DB_URL = f"postgresql://postgres:{password}@localhost:5432/us_equities_historical"
TABLE_NAME = "us_equities"
```

```python
FIELDS_MAP = {
    "f12": "ticker", "f14": "name", "f17": "open", "f15": "high",
    "f16": "low", "f2": "close", "f4": "change", "f3": "pct_chg",
    "f18": "prev_close", "f6": "turnover", "f5": "volume",
    "f20": "mkt_cap", "f13": "market", "f24": "ytd_pct_chg", "f115": "pe_ttm"
}

def get_api_url(pn, pz=100):
    fields = ",".join(FIELDS_MAP.keys())
    return (
        f"https://62.push2delay.eastmoney.com/api/qt/clist/get?"
        f"pn={pn}&pz={pz}&po=1&np=1&ut=bd1d9ddb04089700cf9c27f6f7426281"
        f"&fltt=2&invt=2&fid=f3&fs=m:105,m:106,m:107&fields={fields}"
    )

def fetch_and_save_all():
    # A. 初始化資料庫與日期
    engine = create_engine(DB_URL, pool_pre_ping=True)
    metadata = MetaData()
    # 建議設為固定日期或動態日期
    target_date = (datetime.now() - timedelta(days=1)).strftime('%Y-%m-%d')
    
    try:
        target_table = Table(TABLE_NAME, metadata, autoload_with=engine)
    except Exception as e:
        print(f"🚨 無法載入資料庫表 {TABLE_NAME}：{e}")
        return

    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
        "Connection": "close"
    }
    
    # B. 獲取總頁數
    try:
        init_resp = requests.get(get_api_url(pn=1, pz=1), headers=headers, timeout=15)
        total_count = init_resp.json()['data']['total']
        pz = 100
        total_pages = (total_count // pz) + (1 if total_count % pz > 0 else 0)
        print(f"📊 總數: {total_count} | 預計頁數: {total_pages} | 日期: {target_date}")
    except Exception as e:
        print(f"🚨 初始化請求失敗：{e}")
        return

    # C. 分頁循環抓取
    for pn in range(1, total_pages + 1):
        success = False
        retries = 0
        
        while not success and retries < 5:
            try:
                url = get_api_url(pn, pz)
                resp = requests.get(url, headers=headers, timeout=20)
                resp.raise_for_status()
                
                data = resp.json()
                stocks = data.get("data", {}).get("diff", [])
                
                if stocks:
                    # --- 1. 數據清洗 ---
                    df = pd.DataFrame(stocks)
                    df = df[FIELDS_MAP.keys()].rename(columns=FIELDS_MAP)
                    df['date'] = target_date
                    df = df.replace("-", None)
                    
                    # --- 2. Python 端去重 (防止 CardinalityViolation) ---
                    # 確保這一頁 100 筆資料中沒有重複的 (date, ticker)
                    df = df.drop_duplicates(subset=['date', 'ticker'], keep='first')
                    
                    # --- 3. 核心 Upsert 寫入邏輯 ---
                    data_list = df.to_dict(orient='records')
                    
                    with engine.begin() as conn:
                        # 建立插入語句
                        stmt = pg_insert(target_table).values(data_list)
                        
                        # 定義衝突處理 (ON CONFLICT DO UPDATE)
                        update_dict = {
                            c.name: c for c in stmt.excluded 
                            if c.name not in ['date', 'ticker']
                        }
                        
                        upsert_stmt = stmt.on_conflict_do_update(
                            index_elements=['date', 'ticker'],
                            set_=update_dict
                        )
                        
                        conn.execute(upsert_stmt)
                    
                    print(f"✅ [第 {pn}/{total_pages} 頁] 同步完成 (含更新/去重)")
                    success = True
                    time.sleep(random.uniform(0.6, 1.5))
                else:
                    print(f"⚠️ [第 {pn} 頁] 無數據，重試中...")
                    retries += 1
                    time.sleep(3)
                    
            except (requests.exceptions.ConnectionError, requests.exceptions.ChunkedEncodingError):
                retries += 1
                wait_time = retries * 10
                print(f"❌ 連線中斷，等待 {wait_time} 秒後重試...")
                time.sleep(wait_time)
            except Exception as e:
                retries += 1
                print(f"❌ 發生錯誤：{e}")
                time.sleep(3)

    print("\n🎉 任務結束，所有數據已安全同步。")
```

```python
if __name__ == "__main__":
    fetch_and_save_all()
```

```python
### Manual insert by page 

import requests
import pandas as pd
from sqlalchemy import create_engine, MetaData, Table
from sqlalchemy.dialects.postgresql import insert as pg_insert

# --- 配置 ---
DB_URL = f"postgresql://postgres:{password}@localhost:5432/us_equities_historical"
TABLE_NAME = "us_equities"
TARGET_PAGE = 10
TARGET_DATE = "2026-05-06" 

FIELDS_MAP = {
    "f12": "ticker", "f14": "name", "f17": "open", "f15": "high",
    "f16": "low", "f2": "close", "f4": "change", "f3": "pct_chg",
    "f18": "prev_close", "f6": "turnover", "f5": "volume",
    "f20": "mkt_cap", "f13": "market", "f24": "ytd_pct_chg", "f115": "pe_ttm"
}

def manual_upsert():
    engine = create_engine(DB_URL)
    metadata = MetaData()

    # A. 抓取數據
    print(f"📡 正在抓取第 {TARGET_PAGE} 頁數據...")
    fields = ",".join(FIELDS_MAP.keys())
    url = (
        f"https://62.push2delay.eastmoney.com/api/qt/clist/get?"
        f"pn={TARGET_PAGE}&pz=100&po=1&np=1&ut=bd1d9ddb04089700cf9c27f6f7426281"
        f"&fltt=2&invt=2&fid=f3&fs=m:105,m:106,m:107&fields={fields}"
    )
    
    try:
        resp = requests.get(url, timeout=15)
        stocks = resp.json().get("data", {}).get("diff", [])
        
        if not stocks:
            print("❌ 未抓取到數據。")
            return

        # --- B. 數據整理與「雙重去重」 ---
        df = pd.DataFrame(stocks).rename(columns=FIELDS_MAP)
        df['date'] = TARGET_DATE
        df = df.replace("-", None)

        # 核心修復：防止同一批數據中出現重複的 (date, ticker)
        # 這是為了解決 CardinalityViolation
        initial_len = len(df)
        df = df.drop_duplicates(subset=['date', 'ticker'], keep='first')
        if len(df) < initial_len:
            print(f"💡 發現並移除了 {initial_len - len(df)} 筆 API 回傳的重複項。")

        data_list = df.to_dict(orient='records')

        # --- C. 執行防彈式 Upsert 寫入 ---
        print(f"💾 正在執行 Upsert 寫入表 {TABLE_NAME}...")
        
        # 反射加載表結構
        target_table = Table(TABLE_NAME, metadata, autoload_with=engine)
        
        with engine.begin() as conn:
            # 1. 準備插入語句
            stmt = pg_insert(target_table).values(data_list)
            
            # 2. 處理衝突：當 (date, ticker) 已存在於資料庫時，更新其他欄位
            # stmt.excluded 代表即將寫入的新資料
            update_dict = {
                c.name: c for c in stmt.excluded 
                if c.name not in ['date', 'ticker'] 
            }
            
            upsert_stmt = stmt.on_conflict_do_update(
                index_elements=['date', 'ticker'], 
                set_=update_dict
            )
            
            conn.execute(upsert_stmt)
            
        print(f"✅ PN {TARGET_PAGE} 處理成功！(含自動去重與更新)")

    except Exception as e:
        print(f"🚨 發生錯誤: {e}")

if __name__ == "__main__":
    manual_upsert()
```
